```python
import os
import pandas as pd
from pathlib import Path

folder = Path("/mnt/f/readyparams/param_csvs")

# Dictionary to store concatenated data per species
species_data = {}

for file in folder.glob("*.csv"):
    fname = file.stem  # e.g. "anaxyrus_americanus_subset0"
    
    # Extract species name before "_subset"
    species = fname.split("_subset")[0]
    
    # Read CSV
    df = pd.read_csv(file)
    
    # Append to species group
    if species not in species_data:
        species_data[species] = []
    species_data[species].append(df)

# Now compute unique counts
results = {}

for species, dfs in species_data.items():
    combined = pd.concat(dfs, ignore_index=True)
    unique = combined.drop_duplicates()
    results[species] = len(unique)

# Print results
print("Number of Unique records")
for species, count in results.items():
    print(f"{species}: {count}")
```
